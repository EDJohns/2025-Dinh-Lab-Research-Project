## This script subsamples training data provided by cHL MAPS dataset

```python
import pandas as pd
import random
import os
import numpy as np
import json

os.chdir('..')
os.getcwd()

maps_train = pd.read_csv('./data/cell_phenotyping/train.csv')
```

```python
def sample_cell_indices(df: pd.DataFrame, N: int, current_random_seed: int) -> list:
    """
    Samples N indices per cell label from a DataFrame.
    """
    cell_label_column = df.columns[-1]
    sampled_indices_for_n = []

    # Group by the cell_label column and iterate through each group
    for _, group in df.groupby(cell_label_column):
        # Get the original indices for the current class
        group_indices = list(group.index)
        random.seed(current_random_seed)
        # Sample N indices from this class (or fewer if not enough available)
        sampled_from_class = random.sample(group_indices, k=min(len(group_indices), N))

        # Add these sampled indices to our overall list for this N
        sampled_indices_for_n.extend(sampled_from_class)

    return sampled_indices_for_n


final_sampled_results = {}

# Loop over different numbers of samples (N)
for n_samples_per_class in [2, 5, 10, 15, 20, 30, 50, 100, 200, 500]:
    n_key = f'sampled_{n_samples_per_class}'
    final_sampled_results[n_key] = {}  # Initialize dictionary for this 'n_key'

    # Loop for resampling (5 folds)
    num_resamples = 5
    for fold_num in range(num_resamples):
        # Set random seed for reproducible sampling for this specific fold and N
        current_random_seed = 42 + n_samples_per_class + fold_num
        
        # Call the modified sampling function
        sampled_indices = sample_cell_indices(maps_train, n_samples_per_class, current_random_seed)
        
        # Store the sampled indices under the appropriate resample key
        resample_key = f'resample_{fold_num + 1}'
        final_sampled_results[n_key][resample_key] = sampled_indices
```

```python
# write out structure of sampled cells as json
with open('./data/resample/resampled_indices.json', 'w') as f:
    json.dump(final_sampled_results, f, indent=4) 
```
