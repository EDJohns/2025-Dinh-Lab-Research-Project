```python
maps_pr = []
maps_path = '/Users/jabrand2/Desktop/MAPS'
os.listdir(maps_path)

for f in os.listdir(maps_path):
    if 'pr_curve_data' in os.path.join(maps_path, f):
        fle = os.path.join(maps_path, f)
        print(fle)
        pr = pd.read_csv(fle, low_memory = False)
        pr['model'] = 'MAPS'
        maps_pr.append(pr)
```

```python
rf_pr = []
rf_path = '/Users/jabrand2/Desktop/fluoroforest'
os.listdir(rf_path)

for f in os.listdir(rf_path):
    if 'pr_curve_data' in os.path.join(rf_path, f):
        fle = os.path.join(rf_path, f)
        print(fle)
        pr = pd.read_csv(fle, low_memory = False)
        pr['model'] = 'RandomForest'
        rf_pr.append(pr)
```

```python
merged_pr = pd.concat([pd.concat(rf_pr, axis = 0), 
                       pd.concat(maps_pr, axis = 0)])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import seaborn as sns
from sklearn.metrics import PrecisionRecallDisplay

# Filter for per-class data and for a single fold (Fold 1)
df_filtered = merged_pr[
    (merged_pr['curve_type'] == 'per_class') & 
    (merged_pr['fold'] == 1)
].copy()

# Get unique models and classes
models = sorted(df_filtered['model'].unique())
class_names = sorted(df_filtered['class_name'].unique())
num_classes = len(class_names)
num_models = len(models)

# Set up the two-subplot grid
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 6))
axes = axes.flatten()

# Define a color palette for the classes
class_palette = cm.get_cmap('tab20', num_classes)

# Iterate through each model and plot on a separate subplot
for i, model_name in enumerate(models):
    ax = axes[i]
    
    # Iterate through each class to plot its curve
    for j, class_name in enumerate(class_names):
        # Filter the dataframe for the specific combination
        plot_data = df_filtered[
            (df_filtered['model'] == model_name) &
            (df_filtered['class_name'] == class_name)
        ].copy()
        
        if not plot_data.empty:
            precision_points = plot_data['precision'].values
            recall_points = plot_data['recall'].values
            
            # Use PrecisionRecallDisplay to plot the curve
            display = PrecisionRecallDisplay(
                precision=precision_points,
                recall=recall_points
            )
            
            color = class_palette(j)
            label = f"{class_name}"
            
            display.plot(ax=ax, name=label, color=color, alpha=0.8)
    
    # Customize the subplot
    ax.set_title(f'Precision-Recall Curves for {model_name}', fontsize=14)
    ax.set_xlabel('Recall')
    ax.set_ylabel('Precision')
    # ax.legend(loc="lower left", fontsize=0, handlelength=1.5, frameon=False)
    ax.get_legend().remove()
    ax.grid(False)
    ax.set_xlim([0.0, 1.0])
    ax.set_ylim([0.0, 1.05])

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import auc

df_per_class = merged_pr[merged_pr['curve_type'] == 'per_class'].copy()

# Group by model, class_name, and fold to isolate each unique curve
grouped_data = df_per_class.groupby(['model', 'class_name', 'fold'])

# Create an empty list to store the calculated AP for each curve
ap_values = []

# Iterate through each unique curve to calculate the AUC
for name, group in grouped_data:
    model_name, class_name, fold = name
    
    # Check if the required columns exist in the group before proceeding
    if 'recall' in group.columns and 'precision' in group.columns:
        # Calculate the Area Under the Curve (AUC) from precision and recall
        # The 'auc' function requires x (recall) to be sorted in ascending order.
        group_sorted = group.sort_values(by='recall')
        
        recall_points = group_sorted['recall'].values
        precision_points = group_sorted['precision'].values
        
        # Check for sufficient data points before calculation
        if len(recall_points) > 1 and len(precision_points) > 1:
            average_precision = auc(recall_points, precision_points)
            ap_values.append({'model': model_name, 'class_name': class_name, 'average_precision': average_precision})

# Create a new DataFrame from the calculated AP values
ap_df = pd.DataFrame(ap_values)

# Group the new DataFrame by class and model to get the average AP across all folds
comparison_table = ap_df.groupby(['class_name', 'model'])['average_precision'].mean().unstack()

# Format the table for display
comparison_table = comparison_table.round(4)

# Print the comparison table
print("Average Precision (AP) by Class:")
print(comparison_table)
```

```python

# Convert the table to a long-format DataFrame for plotting
plot_data = comparison_table.reset_index().melt(id_vars='class_name', var_name='model', value_name='average_precision')

# Define the custom color palette
custom_palette = {
    'RandomForest': sns.color_palette('tab20')[0],  # A blue from the tab20 palette
    'MAPS': sns.color_palette('tab20')[2]  # An orange from the tab20 palette
}

# Create the bar plot with the custom palette
plt.figure(figsize=(10, 6))
sns.barplot(
    data=plot_data,
    x='class_name',
    y='average_precision',
    hue='model',
    alpha = 0.8,
    palette=custom_palette
)
# Customize the plot
plt.title('Average Precision (AP) Comparison by Class and Model', fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.xlabel('Class Name', fontsize=12)
plt.ylabel('Average Precision (AP)', fontsize=12)
plt.ylim(0, 1)
plt.legend(title='Model', loc = 'upper left')
plt.tight_layout()

# Display the plot
plt.show()
```
