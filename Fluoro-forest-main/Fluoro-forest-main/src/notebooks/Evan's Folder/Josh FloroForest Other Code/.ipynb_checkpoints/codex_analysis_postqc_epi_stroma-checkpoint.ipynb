```python
# data needs to be rescaled at per core level for each 
og_scale = strepi.layers["scale_data"] # to conirm changes after rescaling
log_norm = strepi.layers["log_data"]

log_df = pd.DataFrame(log_norm).assign(core = strepi.obs["core"].tolist())

# Get a list of all numeric columns, excluding the grouping column
numeric_columns = log_df.select_dtypes(include=[np.number]).columns.tolist()
numeric_columns = [col for col in numeric_columns if col != 'core']

print(numeric_columns)

# Define a function to calculate z-scores
def zscore(group):
    return (group - group.mean()) / group.std()

# Apply z-score normalization to each numeric column within each group
df_zscore = log_df.groupby('core')[numeric_columns].transform(zscore)

# Add the grouping column back to the result
df_zscore['core'] = log_df['core']
```

```python
def find_strings_with_substrings(main_list, include_substrings, exclude_substrings):
    return [
        item for item in main_list 
        if any(include in item for include in include_substrings)
        and not any(exclude in item for exclude in exclude_substrings)
    ]

incl = ["Cell"]
excl = ["DAPI", "CD56", "CD3e", "LAG3", "Granzyme", "CD68", "CD14", "CD163", "FOXP3", "CD4", "CD20", "CD21", "PD.1", "ICOS", "CD8"]
result = find_strings_with_substrings(strepi.var_names, include_substrings = incl, exclude_substrings = excl)
print(result)

pca_features = strepi.var_names.isin(result)
sc.tl.pca(strepi, layer = 'scale_data', mask_var = pca_features)
```

```python
def highlight_cluster(adata, core, resolution, cluster, ignore_warnings = False):
     
    if ignore_warnings:
         warnings.filterwarnings("ignore")
         
    sset = adata.obs.loc[adata.obs['core'] == core,:]
    sset['Centroid.Y.um'] = -1*sset['Centroid.Y.um']

    # Get all unique cluster labels
    all_labels = adata.obs[resolution].unique()

    # Create color mapping for all labels
    cluster_dict = {label: "blue" if label == cluster else "black" for label in all_labels}
    
    # Map colors to the subset
    color_mapping = [cluster_dict[label] for label in sset[resolution]]
    
    # Set point sizes
    sset['size'] = [2 if color == "blue" else 1 for color in color_mapping]

    plt.figure(figsize=(7,6))
    plt.grid(False)
    sns.scatterplot(data=sset.sort_values(by='size'), 
                    x='Centroid.X.um', y='Centroid.Y.um',
                    hue=resolution, palette=cluster_dict,
                    size='size', sizes=(10,30), alpha = 1, 
                   edgecolor = "none")
    plt.title(f'Core = {core}, {resolution} = {cluster}')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')


```

```python
# Confirmed to not be real CD45+ population, can be dropped:
highlight_cluster(adata = strepi, core = "N-10", cluster = "2", resolution="leiden_0.6", ignore_warnings = True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KernelDensity
import seaborn as sns

my_var = "SMA_Cell_Mean"
# Extract data for core B-12
sset = strepi[strepi.obs['core'] == 'M-3']

# Get the index of CD45 in var names
feature_idx = np.where(strepi.var_names == my_var)[0][0]

# Extract coordinates and expression values
x = sset.obs['Centroid.X.um'].values
y = -1*sset.obs['Centroid.Y.um'].values
exprs = sset.layers['log_data'][:, feature_idx].flatten()

# Combine coordinates
coords = np.vstack([x, y]).T

# Fit the kernel density model
kde = KernelDensity(bandwidth = 20, kernel='gaussian')
kde.fit(coords, sample_weight=exprs)

# Create a grid for density estimation
x_grid, y_grid = np.meshgrid(np.linspace(x.min(), x.max(), 50),
                             np.linspace(y.min(), y.max(), 50))
xy_grid = np.vstack([x_grid.ravel(), y_grid.ravel()]).T

# Estimate the density on the grid
z = np.exp(kde.score_samples(xy_grid))
z = z.reshape(x_grid.shape)

# Plot
plt.figure(figsize=(10, 8))
plt.pcolormesh(x_grid, y_grid, z, shading='auto', cmap='viridis')
plt.colorbar(label=f'{my_var} Expression Density')
plt.scatter(x, y, c=exprs, s=5, cmap='viridis', alpha=0.5)
plt.xlabel('Centroid X (µm)')
plt.ylabel('Centroid Y (µm)')
plt.title(f'{my_var} Expression Kernel Density Estimate - Core B-12')
plt.show()

```

```python

ex_data = strepi[strepi.obs['core'] == 'D-5']
ix = ex_data.var_names == 'Pan.Cytokeratin_Cell_Mean'
log_panck_exprs = ex_data.layers['log_data'][:,ix]
plot_df = pd.DataFrame({'PanCk':log_panck_exprs.flatten()})

#plt.hist(plot_df) # log_exprs
```

```python
# log expression
panck_pos = plot_df[plot_df['PanCk'] > 2]
panck_neg = plot_df[plot_df['PanCk'] < 2]
panck_pos_sample = panck_pos.sample(frac = 0.3)

fi_df = pd.concat([panck_neg, panck_pos_sample])
fi_df['PanCk'] = stats.zscore(fi_df['PanCk'])
plot_df['PanCk'] = stats.zscore(plot_df['PanCk'])

panck_pos
```
