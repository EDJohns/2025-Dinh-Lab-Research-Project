### Helper Functions

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def plot_avg_expression_heatmap_by_markers(
    adata,
    cell_type_col="cell_type",
    show_markers=None,
    top_n_genes=None,
    figsize=(10, 8),
    zscore=False,
    core_id=None,
    exclude_unannotated=False
):
    """
    Plots a heatmap of average expression per cell type for selected markers or all genes.

    Parameters
    ----------
    adata : AnnData
        Annotated data matrix.
    cell_type_col : str
        Column in adata.obs defining cell types or clusters.
    show_markers : list of str or None
        List of marker names or substrings to include. 
        If None, all genes in adata.var_names will be used.
    top_n_genes : int or None
        If provided, show top variable genes among the selected markers.
    figsize : tuple
        Figure size (width, height).
    zscore : bool
        If True, normalize expression by z-score per gene.
    exclude_unannotated : bool
        If True, exclude cells where adata.obs['annotations'] == 'not_annotated'.
    """

    # Validate inputs
    if cell_type_col not in adata.obs.columns:
        raise ValueError(f"Column '{cell_type_col}' not found in adata.obs")

    obs = adata.obs.copy()

    # Optionally filter to a specific core ID
    if core_id and "core_id" in obs.columns:
        obs = obs.loc[obs["core_id"] == core_id]
        X = adata[obs.index, :].X
    else:
        X = adata.X

    # Optionally exclude unannotated cells
    if exclude_unannotated and "annotations" in obs.columns:
        mask = obs["annotations"].astype(str) != "not_annotated"
        obs = obs.loc[mask]
        X = adata[obs.index, :].X
    # Optionally exclude unannotated, unknown, and other immune cells
    # if exclude_unannotated and "annotations" in obs.columns:
    #     exclude_terms = ["not_annotated", "unknown", "other_immune"]
    #     mask = ~obs["annotations"].astype(str).str.lower().str.strip().isin(exclude_terms)
    #     obs = obs.loc[mask]
    #     X = adata[obs.index, :].X


    # Expression matrix as DataFrame
    df = pd.DataFrame(
        X.toarray() if hasattr(X, "toarray") else X,
        columns=adata.var_names,
        index=obs.index
    )
    df[cell_type_col] = obs[cell_type_col].values

    # If no specific markers are provided, use all genes
    if show_markers is None:
        selected_genes = [g for g in df.columns if g != cell_type_col]
    else:
        # Select genes containing any of the markers
        selected_genes = [
            g for g in df.columns
            if any(marker.lower() in g.lower() for marker in show_markers)
        ]
        selected_genes = [g for g in selected_genes if g != cell_type_col]

    if not selected_genes:
        raise ValueError("No genes found matching the provided marker list or in the dataset.")

    # Sort selected genes alphabetically
    selected_genes = sorted(selected_genes, key=lambda g: g.lower())

    # Subset to selected genes
    df_filtered = df[selected_genes + [cell_type_col]]

    # Average expression per cell type
    avg_expr = df_filtered.groupby(cell_type_col).mean()

    # Optionally limit to top variable genes
    if top_n_genes and top_n_genes < avg_expr.shape[1]:
        top_genes = avg_expr.var(axis=0).nlargest(top_n_genes).index
        avg_expr = avg_expr[top_genes]

    # Optionally z-score normalize
    if zscore:
        avg_expr = (avg_expr - avg_expr.mean()) / avg_expr.std(ddof=0)

    # Transpose for plotting (genes on Y axis)
    avg_expr_T = avg_expr.T

    # Alphabetically sort rows in the heatmap
    avg_expr_T = avg_expr_T.loc[sorted(avg_expr_T.index, key=lambda x: x.lower())]

    # Plot heatmap
    plt.figure(figsize=figsize)
    sns.heatmap(
        avg_expr_T,
        cmap="Blues",
        annot=True,
        fmt=".2f",
        linewidths=0.5,
        cbar_kws={'label': 'Average Expression (scaled)' if zscore else 'Average Expression'}
    )
    plt.title("Average Gene Expression per " + cell_type_col)
    plt.xlabel("Cell Type")
    plt.ylabel("Genes (alphabetical)")
    plt.tight_layout()
    plt.show()




def plot_avg_expression_heatmap_by_type(
    adata,
    cell_type_col="cell_type",
    expression_type="Cell",  # "Nucleus", "Cytoplasm", "Membrane", "Cell"
    top_n_genes=20,
    figsize=(10, 8),
    core_id=None,
    zscore=False,
    exclude_unannotated=False
):
    """
    Plots a heatmap of average expression per cell type for a given expression type 
    (e.g. "Cell", "Nucleus", "Cytoplasm", "Membrane"), with genes on the right-hand side
    and numeric annotations.

    Parameters
    ----------
    adata : AnnData
        Annotated data matrix.
    cell_type_col : str
        Column in adata.obs defining cell types or clusters.
    expression_type : str
        Substring to filter gene/marker names by ("Cell", "Nucleus", "Cytoplasm", "Membrane").
    top_n_genes : int
        Number of top variable genes to plot.
    figsize : tuple
        Figure size (width, height).
    zscore : bool
        If True, normalize expression by z-score per gene.
    exclude_unannotated : bool
        If True, exclude cells where adata.obs['annotations'] == 'not_annotated'.
    """

    # Check input validity
    if cell_type_col not in adata.obs.columns:
        raise ValueError(f"Column '{cell_type_col}' not found in adata.obs")

    # Create a copy of adata.obs to safely modify
    obs = adata.obs.copy()

    # Optional filtering: exclude unannotated cells
    if exclude_unannotated and "annotations" in obs.columns:
        annotated_mask = obs["annotations"].astype(str) != "not_annotated"
        obs = obs.loc[annotated_mask]
        X = adata[annotated_mask, :].X
    else:
        X = adata.X

    # Convert expression matrix to DataFrame
    df = pd.DataFrame(
        X.toarray() if hasattr(X, "toarray") else X,
        columns=adata.var_names,
        index=obs.index
    )

    # Add cell type info
    df[cell_type_col] = obs[cell_type_col].values

    # Filter genes by expression type substring
    selected_genes = [g for g in df.columns if expression_type.lower() in g.lower()]
    selected_genes = [g for g in selected_genes if g != cell_type_col]

    if not selected_genes:
        raise ValueError(f"No genes found containing '{expression_type}' in their names.")

    # Subset expression data
    df_filtered = df[selected_genes + [cell_type_col]]

    # Compute mean expression per cell type
    avg_expr = df_filtered.groupby(cell_type_col).mean()

    # Select top variable genes
    if top_n_genes and top_n_genes < avg_expr.shape[1]:
        top_genes = avg_expr.var(axis=0).nlargest(top_n_genes).index
        avg_expr = avg_expr[top_genes]

    # Optionally z-score each gene
    if zscore:
        avg_expr = (avg_expr - avg_expr.mean()) / avg_expr.std(ddof=0)

    # Transpose: genes on y-axis, cell types on x-axis
    avg_expr_T = avg_expr.T

    # Plot heatmap with numeric annotations
    plt.figure(figsize=figsize)
    sns.heatmap(
        avg_expr_T,
        cmap="Blues",
        annot=True,
        fmt=".2f",
        linewidths=0.5,
        cbar_kws={'label': 'Average Expression (scaled)' if zscore else 'Average Expression'}
    )
    plt.title(f"Average {expression_type} Expression per {cell_type_col}" +
              (" (excluding unannotated cells)" if exclude_unannotated else ""))
    plt.xlabel("Cell Type")
    plt.ylabel(f"{expression_type} Markers")
    plt.tight_layout()
    plt.show()

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

def plot_avg_expression_heatmap_by_type(
    adata,
    cell_type_col="cell_type",
    expression_type="Cell",  # "Nucleus", "Cytoplasm", "Membrane", "Cell"
    top_n_genes=20,
    figsize=(10, 8),
    core_id=None,
    core_col="core_id",  # name of column identifying core
    zscore=False,
    exclude_unannotated=False
):
    """
    Plots a heatmap of average expression per cell type for a given expression type 
    (e.g. "Cell", "Nucleus", "Cytoplasm", "Membrane"), with genes on the right-hand side
    and numeric annotations. Optionally limited to a specific core.

    Parameters
    ----------
    adata : AnnData
        Annotated data matrix.
    cell_type_col : str
        Column in adata.obs defining cell types or clusters.
    expression_type : str
        Substring to filter gene/marker names by ("Cell", "Nucleus", "Cytoplasm", "Membrane").
    top_n_genes : int
        Number of top variable genes to plot.
    figsize : tuple
        Figure size (width, height).
    core_id : str or None
        If provided, only include cells from this core (adata.obs[core_col] == core_id).
    core_col : str
        Column in adata.obs that contains the core identifier.
    zscore : bool
        If True, normalize expression by z-score per gene.
    exclude_unannotated : bool
        If True, exclude cells where adata.obs['annotations'] == 'not_annotated'.
    """

    # --- Validation ---
    if cell_type_col not in adata.obs.columns:
        raise ValueError(f"Column '{cell_type_col}' not found in adata.obs")

    if core_id is not None:
        if core_col not in adata.obs.columns:
            raise ValueError(f"Column '{core_col}' not found in adata.obs")
        adata = adata[adata.obs[core_col] == core_id].copy()
        if adata.n_obs == 0:
            raise ValueError(f"No cells found for core '{core_id}'")

    # --- Exclude unannotated cells if requested ---
    obs = adata.obs.copy()
    if exclude_unannotated and "annotations" in obs.columns:
        annotated_mask = obs["annotations"].astype(str) != "not_annotated"
        adata = adata[annotated_mask].copy()
        obs = adata.obs.copy()

    # Optionally exclude unannotated, unknown, and other immune cells
    # if exclude_unannotated and "annotations" in obs.columns:
    #     exclude_terms = ["not_annotated", "unknown", "other_immune"]
    #     mask = ~obs["annotations"].astype(str).str.lower().str.strip().isin(exclude_terms)
    #     obs = obs.loc[mask]
    #     X = adata[obs.index, :].X


    # --- Build expression DataFrame ---
    X = adata.X
    df = pd.DataFrame(
        X.toarray() if hasattr(X, "toarray") else X,
        columns=adata.var_names,
        index=obs.index
    )
    df[cell_type_col] = obs[cell_type_col].values

    # --- Filter by expression type ---
    selected_genes = [g for g in df.columns if expression_type.lower() in g.lower()]
    selected_genes = [g for g in selected_genes if g != cell_type_col]
    if not selected_genes:
        raise ValueError(f"No genes found containing '{expression_type}' in their names.")

    # --- Compute average expression ---
    avg_expr = df.groupby(cell_type_col)[selected_genes].mean()

    # --- Select top variable genes ---
    if top_n_genes and top_n_genes < avg_expr.shape[1]:
        top_genes = avg_expr.var(axis=0).nlargest(top_n_genes).index
        avg_expr = avg_expr[top_genes]

    # --- Optionally z-score each gene ---
    if zscore:
        avg_expr = (avg_expr - avg_expr.mean()) / avg_expr.std(ddof=0)

    # --- Transpose for plotting ---
    avg_expr_T = avg_expr.T

    # --- Plot heatmap ---
    plt.figure(figsize=figsize)
    sns.heatmap(
        avg_expr_T,
        cmap="Blues",
        annot=True,
        fmt=".2f",
        linewidths=0.5,
        cbar_kws={'label': 'Average Expression (scaled)' if zscore else 'Average Expression'}
    )
    plt.title(
        f"Average {expression_type} Expression per {cell_type_col}"
        + (f" — Core {core_id}" if core_id else "")
        + (" (excluding unannotated)" if exclude_unannotated else "")
    )
    plt.xlabel("Cell Type")
    plt.ylabel(f"{expression_type} Markers")
    plt.tight_layout()
    plt.show()
def cell_type_counts(adata, cell_type_col="annotations", core_id=None):
    """
    Returns the number of cells for each cell type, optionally filtered by core.

    Parameters
    ----------
    adata : AnnData
        Annotated data matrix.
    cell_type_col : str
        Column in adata.obs containing cell type labels.
    core_id : str or None
        If provided, only counts cells from this core (adata.obs['core_id'] == core_id).

    Returns
    -------
    pd.Series
        Counts of each cell type.
    """
    obs = adata.obs.copy()
    
    if cell_type_col not in obs.columns:
        raise ValueError(f"Column '{cell_type_col}' not found in adata.obs")
    
    if core_id is not None:
        if "core_id" not in obs.columns:
            raise ValueError("Column 'core_id' not found in adata.obs")
        obs = obs[obs["core_id"] == core_id]
    
    counts = obs[cell_type_col].value_counts()
    return counts
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Color mapping for specific cell types
cell_type_colors = {
    "Epi": "#1f77b4",        # Blue
    "Myeloid": "#7f7f7f",    # Orange
    "Endothelia": "#2ca02c", # Green
    "Stroma": "#FFDE59",     # Purple
    "CD8_T": "#2ca02c",      # Red
    "CD4_T": "#8c564b",      # Brown
    "Treg": "#d62728",       # Pink
    "Bcell": "#EE8F00"       # Gray
}
# cell_type_colors = {
#     "Epi": "#1f77b4",        # Blue
#     "Myeloid": "#ff7f0e",    # Orange
#     "Endothelia": "#2ca02c", # Green
#     "Stroma": "#9467bd",     # Purple
#     "CD8_T": "#d62728",      # Red
#     "CD4_T": "#8c564b",      # Brown
#     "Treg": "#e377c2",       # Pink
#     "Bcell": "#7f7f7f"       # Gray
# }

def plot_spatial_per_core(
    adata,
    core_ids,
    core_col="core_id",
    cell_type_col="prediction",
    size=10,
    figsize=(10, 8),
    cell_type_colors=None
):
    OTHER_COLOR = "#d9d9d9"
    if cell_type_colors is None:
        cell_type_colors = {}

    for core in core_ids:
        core_mask = adata.obs[core_col] == core
        subset = adata[core_mask]

        if subset.n_obs == 0:
            continue

        # Coordinates
        coords = subset.obsm["spatial"]
        df = pd.DataFrame({
            "x": coords[:, 0],
            "y": coords[:, 1],
            "cell_type": subset.obs[cell_type_col].astype(str)
        })

        # Map colors
        df["color"] = df["cell_type"].map(cell_type_colors).fillna(OTHER_COLOR)

        # Plot
        plt.figure(figsize=figsize)
        plt.scatter(df["x"], df["y"], c=df["color"], s=size, edgecolor="none")

        plt.title(f"Core: {core}")
        plt.xlabel("X coordinate")
        plt.ylabel("Y coordinate")
        plt.gca().invert_yaxis()

        # Invert Y if needed (to match tissue orientation)
        plt.gca().invert_yaxis()

        # Add a legend with color patches
        handles = [
            plt.Line2D([0], [0], marker="o", color="w",
                       label=ct, markerfacecolor=cell_type_colors.get(ct, OTHER_COLOR), markersize=8)
            for ct in sorted(df["cell_type"].unique())
        ]
        plt.legend(handles=handles, title="Cell Type", bbox_to_anchor=(1.05, 1), loc="upper left")

        plt.tight_layout()
        plt.show()

core_ids = ["coreC7"]
plot_spatial_per_core(
    adata=adata_combined,
    core_ids=core_ids,
    core_col="core_id",
    cell_type_col="prediction",
    size=12,
    figsize=(10, 8),
    cell_type_colors=cell_type_colors
)

# core_ids = ["core12"]
# plot_spatial_per_core(
#     adata=adata_combined,
#     core_ids=core_ids,
#     core_col="core_id",
#     cell_type_col="prediction",
#     size=12,
#     figsize=(10, 8),
#     cell_type_colors=cell_type_colors
# )
```

### Spatial Co-localization

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.neighbors import NearestNeighbors

def plot_cell_colocalization_heatmap(adata, cell_type_col="prediction", n_neighbors=10, figsize=(8, 6)):
    """
    Computes and plots a heatmap showing how often different cell types are near each other.

    Parameters
    ----------
    adata : AnnData
        AnnData object containing spatial coordinates in adata.obsm['spatial'].
    cell_type_col : str
        Column in adata.obs specifying cell types.
    n_neighbors : int
        Number of nearest neighbors to consider for co-localization.
    figsize : tuple
        Figure size for the heatmap.
    """

    # Extract coordinates and cell types
    coords = adata.obsm["spatial"]
    cell_types = adata.obs[cell_type_col].astype(str).values
    unique_types = np.unique(cell_types)

    # Compute nearest neighbors in spatial space
    nbrs = NearestNeighbors(n_neighbors=n_neighbors + 1).fit(coords)
    distances, indices = nbrs.kneighbors(coords)

    # Build a colocalization count matrix
    colocal_counts = pd.DataFrame(0, index=unique_types, columns=unique_types, dtype=float)

    for i, neighbors in enumerate(indices):
        # exclude the cell itself (first neighbor is the cell itself)
        neighbors = neighbors[1:]
        this_type = cell_types[i]
        neighbor_types = cell_types[neighbors]
        for nt in neighbor_types:
            colocal_counts.loc[this_type, nt] += 1

    # Normalize by total neighbors per cell type (optional)
    colocal_freq = colocal_counts.div(colocal_counts.sum(axis=1), axis=0)

    # Plot heatmap
    plt.figure(figsize=figsize)
    sns.heatmap(colocal_freq, cmap="viridis", annot=True, fmt=".2f", square=True)
    plt.title(f"Cell–Cell Colocalization (k={n_neighbors})")
    plt.xlabel("Neighbor cell type")
    plt.ylabel("Reference cell type")
    plt.tight_layout()
    plt.show()

    return colocal_freq

colocal_matrix = plot_cell_colocalization_heatmap(
    adata=adata_combined,
    cell_type_col="prediction",
    n_neighbors=10,
    figsize=(8, 6)
)
```

### Z-Score Colocalization Model

    Compute and visualize a Z-score–based cell–cell colocalization heatmap
    to identify statistically enriched or depleted spatial associations
    between cell types within a given tissue core.

    Methodology
    ------------
    This analysis quantifies how frequently different cell types occur
    in close spatial proximity compared to what would be expected
    under a null hypothesis of random spatial organization.

    1. Input data
       - Each observation (row) in the AnnData object represents a single cell.
       - Spatial coordinates are stored in `adata.obsm["spatial"]`.
       - Each cell has an associated categorical label (e.g. cell type)
         stored in `adata.obs[cell_type_col]`.
       - The dataset is typically restricted to one core or region of interest.

    2. Nearest-neighbor graph construction
       - For each cell, its *k* nearest neighbors are identified in the 2D
         spatial coordinate space using Euclidean distance.
       - The number of neighbors (*k*, given by `n_neighbors`) controls
         the local scale of spatial relationships considered.

    3. Observed co-occurrence matrix
       - For each reference cell type *A*, we count how many of its
         neighboring cells belong to each other type *B*.
       - These counts are aggregated across all cells and normalized by the
         total number of neighbors for that reference cell type,
         yielding an observed co-occurrence frequency matrix *O(A,B)*.

    4. Null model (random expectation)
       - To estimate the frequency expected by random chance, we repeatedly
         permute the cell type labels across all spatial coordinates
         (`n_permutations` times), breaking any true spatial relationships.
       - For each permutation, a co-occurrence matrix is recomputed in the
         same way as above, producing a distribution of random co-occurrence
         frequencies for each cell type pair *A–B*.
       - The mean and standard deviation across permutations define
         the expected frequency (*E(A,B)*) and its variability (*σ(A,B)*).

    5. Z-score computation
       - The deviation of the observed frequency from random expectation is
         standardized as a Z-score:
             Z(A,B) = [ O(A,B) - E(A,B) ] / σ(A,B)
       - Positive Z-scores indicate that cell type *B* is found near
         cell type *A* more often than expected by chance (spatial enrichment).
       - Negative Z-scores indicate underrepresentation (spatial segregation).

    6. Visualization
       - The resulting symmetric Z-score matrix is displayed as a heatmap
         using a diverging color scale (e.g., blue–white–red) centered at zero.
       - Each cell type pair’s Z-score conveys the strength and direction
         of spatial association.
       - The plot title includes the core name and number of neighbors used.

    Biological Interpretation
    -------------------------
    - **Z > 0:** Co-localization enrichment — the two cell types
      preferentially occupy nearby spatial regions, potentially indicating
      interaction, shared microenvironments, or coordinated localization.
    - **Z < 0:** Spatial exclusion — the two cell types avoid proximity,
      suggesting compartmentalization or distinct tissue niches.
    - **Z ≈ 0:** Random mixing — no statistically meaningful association.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.neighbors import NearestNeighbors

def plot_cell_colocalization_zscore(
    adata,
    cell_type_col="prediction",
    core_name="unknown core",
    n_neighbors=10,
    n_permutations=100,
    figsize=(8, 6),
    random_state=0
):
    """
    Compute a Z-score colocalization heatmap showing which cell types co-localize more
    or less than expected by chance.

    Parameters
    ----------
    adata : AnnData
        AnnData object with spatial coordinates in adata.obsm['spatial'].
    cell_type_col : str
        Column in adata.obs specifying cell types.
    core_name : str
        Label for the plot title.
    n_neighbors : int
        Number of nearest neighbors to consider for co-localization.
    n_permutations : int
        Number of random shuffles for estimating expected co-localization.
    figsize : tuple
        Figure size for the heatmap.
    random_state : int
        Random seed for reproducibility.

    Returns
    -------
    zscore_df : pd.DataFrame
        Z-score colocalization matrix.
    """

    core_mask = adata.obs["core_id"] == core_name
    subset = adata[core_mask].copy()

    rng = np.random.default_rng(random_state)

    coords = subset.obsm["spatial"]
    cell_types = subset.obs[cell_type_col].astype(str).values
    unique_types = np.unique(cell_types)

    # Fit nearest neighbors
    nbrs = NearestNeighbors(n_neighbors=n_neighbors + 1).fit(coords)
    distances, indices = nbrs.kneighbors(coords)

    # --- Compute observed counts ---
    obs_counts = pd.DataFrame(0, index=unique_types, columns=unique_types, dtype=float)

    for i, neighbors in enumerate(indices):
        neighbors = neighbors[1:]  # skip self
        this_type = cell_types[i]
        neighbor_types = cell_types[neighbors]
        for nt in neighbor_types:
            obs_counts.loc[this_type, nt] += 1

    # Normalize observed by row sums
    obs_freq = obs_counts.div(obs_counts.sum(axis=1), axis=0).fillna(0)

    # --- Compute expected counts by permutation ---
    perm_freqs = np.zeros((len(unique_types), len(unique_types), n_permutations))

    for p in range(n_permutations):
        shuffled = rng.permutation(cell_types)
        perm_counts = pd.DataFrame(0, index=unique_types, columns=unique_types, dtype=float)

        for i, neighbors in enumerate(indices):
            neighbors = neighbors[1:]
            this_type = shuffled[i]
            neighbor_types = shuffled[neighbors]
            for nt in neighbor_types:
                perm_counts.loc[this_type, nt] += 1

        perm_freqs[:, :, p] = perm_counts.div(perm_counts.sum(axis=1), axis=0).fillna(0).values

    # --- Compute Z-scores ---
    perm_mean = perm_freqs.mean(axis=2)
    perm_std = perm_freqs.std(axis=2)
    zscore = (obs_freq.values - perm_mean) / (perm_std + 1e-9)  # avoid divide-by-zero
    zscore_df = pd.DataFrame(zscore, index=unique_types, columns=unique_types)

    # --- Plot heatmap ---
    plt.figure(figsize=figsize)
    sns.heatmap(
        zscore_df,
        cmap="coolwarm",
        center=0,
        annot=True,
        fmt=".2f",
        square=True,
        cbar_kws={"label": "Z-score"}
    )
    plt.title(f"Cell–Cell Colocalization Z-score ({core_name}, k={n_neighbors})")
    plt.xlabel("Neighbor cell type")
    plt.ylabel("Reference cell type")
    plt.tight_layout()
    plt.show()

    return zscore_df

zscore_matrix = plot_cell_colocalization_zscore(
    adata=adata_combined,                # or your adata_combined slice per core
    cell_type_col="prediction",
    core_name="coreN12",
    n_neighbors=6,
    n_permutations=10,          # increase to 500+ for more stability
    figsize=(8, 6)
)
```

```python
import squidpy as sq
import warnings
warnings.filterwarnings("ignore")

# Color mapping for specific cell types
cell_type_colors = {
    "Epi": "#1f77b4",        # Blue
    "Myeloid": "#ff7f0e",    # Orange
    "Endothelia": "#2ca02c", # Green
    "Stroma": "#9467bd",     # Purple
    "CD8_T": "#d62728",      # Red
    "CD4_T": "#8c564b",      # Brown
    "Treg": "#e377c2",       # Pink
    "Bcell": "#7f7f7f"       # Gray
}

def plot_spatial_per_core(
    adata,
    core_ids,
    core_col="core_id",
    cell_type_col="prediction",
    size=10,
    figsize=(10, 8),
    cell_type_colors=None
):
    """
    Plot spatial scatter plots per core with proper colors.

    Parameters
    ----------
    adata : AnnData
        The combined AnnData object containing multiple cores.
    core_ids : list of str
        List of core IDs to plot (should match adata.obs[core_col]).
    core_col : str
        Column in adata.obs representing core IDs.
    cell_type_col : str
        Column in adata.obs with cell type annotations.
    patient_col : str
        Column in adata.obs with patient ID.
    response_col : str
        Column in adata.obs with response status.
    size : int
        Marker size in scatter plot.
    figsize : tuple
        Size of the plot.
    cell_type_colors : dict
        Dictionary mapping cell types to colors.
    """

    OTHER_COLOR = "#d9d9d9"  # default for unknown types
    if cell_type_colors is None:
        cell_type_colors = {}

    for core in core_ids:
        core_mask = adata.obs[core_col] == core
        subset = adata[core_mask]

        # Skip if no cells in this core
        if subset.n_obs == 0:
            continue

        # Convert cell type to categorical
        subset.obs[cell_type_col] = subset.obs[cell_type_col].astype("category")
        category_order = subset.obs[cell_type_col].cat.categories.tolist()

        # Generate color list in the correct order
        color_list = [cell_type_colors.get(ct, OTHER_COLOR) for ct in category_order]
        subset.uns[f"{cell_type_col}_colors"] = color_list

        # Plot and get the axes object
        ax = sq.pl.spatial_scatter(
            subset,
            color=cell_type_col,
            size=size,
            shape=None,
            title=f"Core: {core}",
            figsize=figsize,
            legend_loc="right margin",
            return_ax=True  # <- This is key
        )
        
        # Customize axes
        ax.set_xlabel("X coordinate")
        ax.set_ylabel("Y coordinate")
        ax.set_xticks(ax.get_xticks())  # reinitialize ticks
        ax.set_yticks(ax.get_yticks())
        for spine in ax.spines.values():
            spine.set_visible(True)
        
        # Show axes ticks (sometimes hidden by default)
        ax.tick_params(left=True, bottom=True, labelleft=True, labelbottom=True)
        
        plt.show()

            
core_ids = ["core7","core 12"]  # list of cores you want to plot
plot_spatial_per_core(
    adata=adata_combined,
    core_ids=core_ids,
    core_col="core_id",
    cell_type_col="prediction",
    size=10,
    figsize=(12, 10),
    cell_type_colors=cell_type_colors
)
```
