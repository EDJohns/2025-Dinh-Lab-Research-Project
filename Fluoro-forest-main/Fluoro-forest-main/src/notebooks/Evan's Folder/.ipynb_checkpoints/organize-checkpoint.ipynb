```python
import os
import shutil
import pandas as pd
from pathlib import Path
```

```python
unsorted = Path('unsorted')
main_folder = Path('sorted')
os.makedirs(main_folder, exist_ok = True)
expression_df = pd.read_csv(unsorted / 'codex_log2.csv', low_memory = False, index_col = 0)
cell_meta = pd.read_csv(unsorted / 'cell_meta.csv')

# can replace this with glob method
ome_tif_files = []
for entry_name in os.listdir(unsorted):
    full_path = unsorted / entry_name
    if full_path.is_file() and entry_name.endswith('.ome.tif'):
        ome_tif_files.append(full_path)
```

```python
cell_meta_df = cell_meta.loc[:,['Object.ID', 'Centroid.X.um', 'Centroid.Y.um', 'core']].set_index('Object.ID')
```

```python
# get each file in the array and get the core name 
for ome_tif_path in ome_tif_files:
    core_name = ome_tif_path.name.split('.')[0]
    core_dir = main_folder / core_name
    os.makedirs(core_dir, exist_ok = True)
    
    os.makedirs(core_dir / 'annotations', exist_ok = True)
    os.makedirs(core_dir / 'expression', exist_ok = True)
    os.makedirs(core_dir / 'image', exist_ok = True)
    os.makedirs(core_dir / 'model', exist_ok = True)
    os.makedirs(core_dir / 'segmentation', exist_ok = True)
    
    #move the images
 
    destination_image_path = core_dir/'image'/ome_tif_path.name
    shutil.move(ome_tif_path, destination_image_path)
    
    #move the geojson 
    geojson_path_unsorted = unsorted / f"{core_name}.geojson"
    destination_geojson_path = core_dir / 'segmentation' / geojson_path_unsorted.name
    shutil.move(geojson_path_unsorted, destination_geojson_path)
    
    filtered_meta_df = cell_meta_df[cell_meta_df['core'] == core_name].copy()
    
    meta_output_path = core_dir / 'expression' / f"{core_name}_meta.csv"
    filtered_meta_df.to_csv(meta_output_path, index = True) 

    core_object_ids = filtered_meta_df.index

    filtered_expression_df = expression_df[expression_df.index.isin(core_object_ids)].copy()
    expression_output_path = core_dir / 'expression' / f"{core_name}_expression.csv"
    filtered_expression_df.to_csv(expression_output_path, index = True)
```
