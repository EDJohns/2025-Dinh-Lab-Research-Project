```python
import scanpy as sc
import squidpy as sq
import pandas as pd
import numpy as np

# Load files on Evan's Laptop
# expr_orig = pd.read_csv(r"C:\Users\evanj\OneDrive\Documents\expression.csv", index_col=0)
# expr=expr_orig.transpose()
# metadata = pd.read_csv(r"C:\Users\evanj\OneDrive\Documents\metadata.csv", index_col=0)
# umap = pd.read_csv(r"C:\Users\evanj\OneDrive\Documents\umap.csv", index_col=0)

#Load files on Lab computer
expr_orig = pd.read_csv(r"C:\Users\ejohns\Documents\Shapiro Data Files\expression.csv", index_col=0)
expr=expr_orig.transpose()
metadata = pd.read_csv(r"C:\Users\ejohns\Documents\Shapiro Data Files\metadata.csv", index_col=0)
umap = pd.read_csv(r"C:\Users\ejohns\Documents\Shapiro Data Files\umap.csv", index_col=0)

# Create AnnData object
adata = sc.AnnData(X=expr.values)

# Assign metadata
adata.obs = metadata
adata.var_names = expr.columns
adata.obs_names = expr.index

# Add spatial coordinates and UMAP to .obsm
# adata.obsm["spatial"] = metadata[['x_FOV_px', 'y_FOV_px']].values  # adjust if needed
adata.obsm["spatial"] = metadata[['x_FOV_px']].assign(y_FOV_px = -metadata['y_FOV_px']).values
adata.obsm["X_umap"] = umap.values
```

```python
# Distance Functions
import squidpy as sq
import pandas as pd

cell_ID_1="c_1_1_1"
cell_ID_2="c_1_1_212"

def get_xy_coords(cell_ID):
    x_coord=adata.obs.loc[cell_ID, "x_FOV_px"]
    y_coord=adata.obs.loc[cell_ID, "y_FOV_px"]
    # x=adata.obs.loc[cell_ID, "merged_annot_cluster"]
    # display(x)
    return{
        "X":x_coord,
        "Y":y_coord
    }
def find_distance(cell_ID_1,cell_ID_2,unit_conversion=0.0168):
    from math import sqrt
    # Get Cell Locations
    cell_1_xy=get_xy_coords(cell_ID_1)
    cell_2_xy=get_xy_coords(cell_ID_2)

    # Distance Function
    Distance = sqrt((cell_2_xy["X"] - cell_1_xy["X"])**2 + (cell_2_xy["Y"] - cell_1_xy["Y"])**2)

    return{
            "Distance":Distance
        }

result=find_distance(cell_ID_1,cell_ID_2,unit_conversion=0.0168)["Distance"]
display(result)
```

```python
# Goals
# (1) create list of pmn cells
# (2) create distance function 
# (3) create matrix of pmn distance to each T-cell (make it modular for other cell types)
import my_functions

sample_ID="c_1_1"

pmn_list=my_functions.pmn_counter(adata,sample_ID)['PMN Names']
cell_distance_type="CD4+T_cells"

for pmn in pmn_list:
    # Endpoint to iterate over the while loop
    endpoint=False

    # Loops through all cells in the sample
    cell_num=1
    kill_count=0 # Number of failed calls
    while endpoint==False:
        cell_num+=1
        cell_id=f"{sample_ID}_{cell_num}"

        # If the other cells is a pmn skip
        if cell_id in pmn_list:
            continue
            
        # We count up by 1 each time but not every cell exits so we use the try function
        try:
            cell_type=adata.obs.loc[cell_id, "merged_annot_cluster"]
            if cell_distance_type==cell_type:
                distance=find_distance(pmn,cell_id)["Distance"]
                display(f"cell1:{pmn} cell 2:{cell_id} Distance:{distance}")
        except:
            kill_count+=1

        if kill_count>500:
            endpoint=True
```
