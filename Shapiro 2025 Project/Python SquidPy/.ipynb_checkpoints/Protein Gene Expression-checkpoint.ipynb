## Protein Gene Expression Analysis 

Moving protein expression analysis here. This finds the average gene expression by cell type and divides into groups based on proximities with other cells. 

**Note to Specify if two are going 1 cell comparison (2 groups) or 2 cell comparison (4 groups)**

```python
import my_functions
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Initialize storage. Must be done outside the loop for best results.
responder_dfs = []
nonresponder_dfs = []
combined_sample_avg = []  
stats_NRvR = []

# Cell options: Neutrophils, CD8+T_cells, Tumor_cells, Treg
# Parameters
distance_threshold = 15
target_cell_type = "CD8+T_cells"
cell_type_1 = "Neutrophils"
cell_type_2 = "Tumor_cells"
decimal_threshold = 0.0025

for i in range(1, 4):  
    for j in range(1, 26): 
        sample_id = f"c_{i}_{j}"
        cell_decimal=cell_deciminal=my_functions.matching_cell_list(adata,sample_id,target_cell_type)["Decimal Count"]
        if cell_decimal<decimal_threshold:
            print(f"Sample c_{i}_{j} Skipped due to low cell count")
            continue
        try:
            responder_status = my_functions.get_sample_info(adata, sample_id)['Response']  # 'R' or 'NR'

            # Try to extract the sample subset
            adata_sample = adata[adata.obs.cell_ID.str.contains(sample_id)].copy() #adata[adata.obs['sample'] == sample_id]  # Adjust key if needed

            result = my_functions.get_interaction_status_v2(
                adata_sample,
                sample_id,
                distance_threshold,
                target_cell_type,
                cell_type_1=cell_type_1
                #,cell_type_2=cell_type_2
            )
            interaction_matrix = result["Interaction Matrix"]

            # Extract protein expression matrix
            expr = adata_sample.X  # Or use .layers['protein'] if applicable
            expr_df = pd.DataFrame(
                expr.toarray() if hasattr(expr, 'toarray') else expr,
                index=adata_sample.obs_names,
                columns=adata_sample.var_names
            )
            expr_df['Interaction_Status'] = interaction_matrix['Interaction_Status']

            #display(interaction_matrix['Interaction_Status'])
            sample_avg=expr_df.groupby('Interaction_Status').mean()
            sample_avg['sample_id']=sample_id
            sample_avg['responder_status']=responder_status

            # Get Stats Row for NR versus R
            stats_NRvR_row = pd.DataFrame(expr_df.mean(numeric_only=True)).transpose()
            stats_NRvR_row["sample_id"]=sample_id
            stats_NRvR_row["responder_status"]=responder_status

            # Initalize Group Matrices 
            combined_sample_avg.append(sample_avg)  # Append to list
            stats_NRvR.append(stats_NRvR_row)

            # Append to appropriate list
            if responder_status == 'R':
                responder_dfs.append(expr_df)
            elif responder_status == 'NR':
                nonresponder_dfs.append(expr_df)

            #Prints Status to Track
            print(f"Sample c_{i}_{j} complete")
            
        except Exception as e:
            print(f"Skipping sample {sample_id}: {e}")
            continue

combined_sample_avg = pd.concat(combined_sample_avg).reset_index()
stats_NRvR = pd.concat(stats_NRvR).reset_index()
display(stats_NRvR)

# Combine all samples by group
responder_combined = pd.concat(responder_dfs)
nonresponder_combined = pd.concat(nonresponder_dfs)

# Function to plot from combined df
def plot_grouped_heatmap_from_df(expr_df, title, vmax=8):
    group_counts = expr_df['Interaction_Status'].value_counts()
    grouped_expr = expr_df.groupby('Interaction_Status').mean()
   # grouped_expr.index = [f"{g} (n={group_counts[g]})" for g in grouped_expr.index]
    grouped_expr = grouped_expr.sort_index()
    plt.figure(figsize=(max(10, len(grouped_expr.columns) * 0.5+2), len(grouped_expr) * 0.6 + 2))
    sns.heatmap(grouped_expr, cmap="Spectral", annot=True, vmin=0, vmax=vmax)
    plt.title(title)
    plt.ylabel('Interaction Group')
    plt.xlabel('Protein Marker')
    plt.tight_layout()
    plt.show()

# Plot for responders
plot_grouped_heatmap_from_df(responder_combined, title="Responders - PMN Protein Expression by Interaction Group")

# Plot for non-responders
plot_grouped_heatmap_from_df(nonresponder_combined, title="Non-Responders - PMN Protein Expression by Interaction Group")
```

### Significance in expression testing:

Trying to answer two questions with this analysis (1) R vs NR signficant difference in PMN, CD8, or Treg expression (2) Interacting vs Non interaction PMN, CD8, Treg expression

#### Non-Responder vs Responder Statistical Signifance

```python
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests
import plotly.express as px
import pandas as pd
import numpy as np

# Separate responders and non-responders
R_samples = stats_NRvR[stats_NRvR['responder_status'] == 'R']
NR_samples = stats_NRvR[stats_NRvR['responder_status'] == 'NR']

# Get all protein marker columns (exclude non-numeric and metadata columns)
exclude_cols = ['sample_id', 'responder_status', 'index']
marker_cols = [col for col in stats_NRvR.columns if col not in exclude_cols and pd.api.types.is_numeric_dtype(stats_NRvR[col])]

# Run Wilcoxon rank-sum test for each marker
results = []

for marker in marker_cols:
    r_vals = R_samples[marker].dropna()
    nr_vals = NR_samples[marker].dropna()

    if len(r_vals) > 0 and len(nr_vals) > 0:
        stat, pval = mannwhitneyu(r_vals, nr_vals, alternative='two-sided')

        # Compute log2 fold change (avoid division by 0)
        log2_fc = r_vals.mean() - nr_vals.mean()

        results.append({
            'Marker': marker,
            'R_mean': r_vals.mean(),
            'NR_mean': nr_vals.mean(),
            'U_stat': stat,
            'p_value': pval,
            'log2_FC': log2_fc
        })
    else:
        results.append({
            'Marker': marker,
            'R_mean': None,
            'NR_mean': None,
            'U_stat': None,
            'p_value': None,
            'log2_FC': None
        })

# Convert to DataFrame
wilcoxon_results_df = pd.DataFrame(results).dropna().sort_values('p_value')

# Optional: Adjust p-values for multiple comparisons (FDR)
pvals = wilcoxon_results_df['p_value'].values
_, fdr_corrected_pvals, _, _ = multipletests(pvals, method='fdr_bh')
wilcoxon_results_df['fdr_p_value'] = fdr_corrected_pvals

# Display sorted results
display(wilcoxon_results_df)

# Step 1: Find the most significant marker by FDR
top_marker = wilcoxon_results_df.sort_values('fdr_p_value').iloc[0]['Marker']
print(f"Most significant marker: {top_marker}")

# Step 2: Create box plot
boxplot_df = stats_NRvR[['responder_status', top_marker]].copy().dropna()
fig = px.box(
    boxplot_df,
    x='responder_status',
    y=top_marker,
    points="all",
    color='responder_status',
    title=f"Expression of {top_marker} by Responder Status",
    labels={'responder_status': 'Responder Status', top_marker: 'Expression'}
)
fig.update_traces(jitter=0.3, marker_opacity=0.6)
fig.update_layout(width=900, height=600)
fig.show()

# Step 3: Volcano Plot
volcano_df = wilcoxon_results_df.dropna(subset=['log2_FC', 'p_value']).copy()
volcano_df['-log10_pval'] = -np.log10(volcano_df['p_value'])

volcano_fig = px.scatter(
    volcano_df,
    x='log2_FC',
    y='-log10_pval',
    text='Marker',
    title=F"{target_cell_type} Volcano Plot: Responders vs Non-Responders",
    labels={'log2_FC': 'log₂ Fold Change (R / NR)', '-log10_pval': '-log₁₀(p-value)'},
    hover_data=['Marker', 'p_value', 'fdr_p_value', 'log2_FC']
)
volcano_fig.update_traces(marker=dict(size=8, opacity=0.7), textposition='top center')
volcano_fig.update_layout(width=900, height=600)
volcano_fig.show()
```

### Interacting vs Non Intecting (Each sample dividided into interacting or non-interacting

```python
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests
import plotly.express as px
import pandas as pd

def plot_interaction_marker_stats(combined_sample_avg, responder_filter=None):
    """
    Creates an interactive boxplot of the most significant marker between Interacting vs Not Interacting groups.
    
    Parameters:
        combined_sample_avg (DataFrame): The input data with expression, interaction, and response info.
        responder_filter (str or None): 'R', 'NR', or None to filter by responder status.
    """
    # Optionally filter by responder status
    if responder_filter in ['R', 'NR']:
        data = combined_sample_avg[combined_sample_avg['responder_status'] == responder_filter]
        print(f"Filtering for responder_status = '{responder_filter}'")
    else:
        data = combined_sample_avg.copy()

    # Ensure only numeric expression columns are used
    exclude_cols = ['Interaction_Status', 'sample_id', 'responder_status']
    expr_cols = [col for col in data.columns if col not in exclude_cols and pd.api.types.is_numeric_dtype(data[col])]

    # Separate groups
    interacting = data[data['Interaction_Status'] == 'Interacting']
    non_interacting = data[data['Interaction_Status'] == 'Not Interacting']

    # Wilcoxon test for each marker
    interaction_results = []
    for marker in expr_cols:
        x = interacting[marker].dropna()
        y = non_interacting[marker].dropna()
        if len(x) > 0 and len(y) > 0:
            stat, p = mannwhitneyu(x, y, alternative='two-sided')
            interaction_results.append({
                'Marker': marker,
                'Interacting_mean': x.mean(),
                'NonInteracting_mean': y.mean(),
                'U_stat': stat,
                'p_value': p
            })

    # Create DataFrame with stats
    interaction_test_df = pd.DataFrame(interaction_results)
    if interaction_test_df.empty:
        print("No valid markers to compare.")
        return
#wilcoxon_results_df = pd.DataFrame(results).sort_values('p_value')
    interaction_test_df['fdr_p_value'] = multipletests(interaction_test_df['p_value'], method='fdr_bh')[1]
    interaction_test_df = interaction_test_df.sort_values('p_value')
    display(interaction_test_df.head(20))

    # Plot most significant marker
    top_marker = interaction_test_df.iloc[0]['Marker']
    print(f"Most significant marker: {top_marker}")

    plot_df = data[['Interaction_Status', top_marker, 'sample_id']].dropna()
    
    fig = px.box(
        plot_df,
        x='Interaction_Status',
        y=top_marker,
        color='Interaction_Status',
        points='all',
        title=f"{target_cell_type} Expression of {top_marker} by Interaction Status",
        labels={top_marker: "Expression Level", 'Interaction_Status': 'Interaction Group'},
        hover_data=['sample_id']
    )

    fig.update_layout(width=900, height=600)
    fig.update_traces(jitter=0.3, marker_opacity=0.5)
    fig.show()

# Example usage:
plot_interaction_marker_stats(combined_sample_avg)         # All samples
plot_interaction_marker_stats(combined_sample_avg, 'R')    # Responders only
plot_interaction_marker_stats(combined_sample_avg, 'NR')   # Non-responders only
```

### Volcano Plot

A volcano plot may also be better way to view this signifcance 

```python
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests
import plotly.express as px
import pandas as pd
import numpy as np

def plot_interaction_marker_stats(combined_sample_avg, responder_filter=None, assume_log=False):
    """
    Creates an interactive boxplot and volcano plot of the most significant marker 
    between Interacting vs Not Interacting groups.
    
    Parameters:
        combined_sample_avg (DataFrame): Input data with expression, interaction, and response info.
        responder_filter (str or None): 'R', 'NR', or None to filter by responder status.
        assume_log (bool): If True, assumes expression data is already log-transformed (e.g., log2).
    """
    # Step 1: Filter by responder status if requested
    if responder_filter in ['R', 'NR']:
        data = combined_sample_avg[combined_sample_avg['responder_status'] == responder_filter]
        print(f"Filtering for responder_status = '{responder_filter}'")
    else:
        data = combined_sample_avg.copy()

    # Step 2: Identify numeric expression columns
    exclude_cols = ['Interaction_Status', 'sample_id', 'responder_status']
    expr_cols = [col for col in data.columns if col not in exclude_cols and pd.api.types.is_numeric_dtype(data[col])]

    # Step 3: Split data into interacting vs. non-interacting groups
    interacting = data[data['Interaction_Status'] == 'Interacting']
    non_interacting = data[data['Interaction_Status'] == 'Not Interacting']

    # Step 4: Run statistical tests
    interaction_results = []
    for marker in expr_cols:
        x = interacting[marker].dropna()
        y = non_interacting[marker].dropna()
        if len(x) > 0 and len(y) > 0:
            stat, p = mannwhitneyu(x, y, alternative='two-sided')

            # Calculate log2 fold change depending on input scale
            if assume_log:
                log2_fc = x.mean() - y.mean()
            else:
                # Avoid division by zero or log of negative
                if x.mean() > 0 and y.mean() > 0:
                    log2_fc = np.log2(x.mean() / y.mean())
                else:
                    log2_fc = np.nan

            interaction_results.append({
                'Marker': marker,
                'Interacting_mean': x.mean(),
                'NonInteracting_mean': y.mean(),
                'U_stat': stat,
                'p_value': p,
                'log2_FC': log2_fc
            })

    # Step 5: Create summary DataFrame
    interaction_test_df = pd.DataFrame(interaction_results)
    if interaction_test_df.empty:
        print("No valid markers to compare.")
        return

    # FDR correction
    interaction_test_df['fdr_p_value'] = multipletests(interaction_test_df['p_value'], method='fdr_bh')[1]

    # Step 6: Box Plot for Top Marker
    top_marker = interaction_test_df.sort_values('p_value').iloc[0]['Marker']
    print(f"Most significant marker: {top_marker}")

    plot_df = data[['Interaction_Status', top_marker, 'sample_id']].dropna()
    fig = px.box(
        plot_df,
        x='Interaction_Status',
        y=top_marker,
        color='Interaction_Status',
        points='all',
        title=f"{target_cell_type} Expression of {top_marker} by Interaction Status",
        labels={top_marker: "Expression Level", 'Interaction_Status': 'Interaction Group'},
        hover_data=['sample_id']
    )
    fig.update_layout(width=900, height=600)
    fig.update_traces(jitter=0.3, marker_opacity=0.5)
    fig.show()

    # Step 7: Volcano Plot
    volcano_df = interaction_test_df.dropna(subset=['log2_FC', 'p_value'])
    volcano_fig = px.scatter(
        volcano_df,
        x='log2_FC',
        y=-np.log10(volcano_df['p_value']),
        text='Marker',
        title=f" {target_cell_type} Volcano Plot: Responder Filter: {responder_filter} Int vs Non-Int",
        labels={'log2_FC': 'log₂ Fold Change', 'y': '-log₁₀(p-value)'},
        hover_data=['Marker', 'p_value', 'log2_FC']
    )
    volcano_fig.update_traces(marker=dict(size=8, opacity=0.7), textposition='top center')
    volcano_fig.update_layout(width=900, height=600)
    volcano_fig.show()

    # Step 8: Display top results
    display(interaction_test_df.sort_values('p_value').head(20))
    
plot_interaction_marker_stats(combined_sample_avg, assume_log=True)               # All samples
plot_interaction_marker_stats(combined_sample_avg, 'R', assume_log=True)          # Responders
plot_interaction_marker_stats(combined_sample_avg, 'NR', assume_log=True)         # Non-responders

```

### Gene Expression Plot Grey Map

This function allows you to specify a protein and it will show a map of it's expression in a light/dark manner. 

Input:
* TMA/Sample Id
* Mapping Gene

Output:
* Greyscale graph of gene expression

```python
import squidpy as sq
import warnings
import my_functions
warnings.filterwarnings("ignore")

cell_ID_like="c_3_7_"
mapping_gene="CD16"

subset = adata[adata.obs.cell_ID.str.contains(cell_ID_like)]
pmn_count = my_functions.pmn_counter(adata, cell_ID_like)["PMN Count"]
patient_id = subset.obs["Patient.ID"].unique()[0]

# Convert to categorical without setting all 12 types
subset.obs["merged_annot_cluster"] = subset.obs["merged_annot_cluster"].astype("category")

# Get gene expression and store in obs
gene_values = subset[:, mapping_gene].X
if hasattr(gene_values, "toarray"):  # handle sparse matrix
    gene_values = gene_values.toarray().flatten()
else:
    gene_values = gene_values.flatten()

subset.obs[f"{mapping_gene}_expression"] = gene_values

# Plot using gene expression as color
sq.pl.spatial_scatter(
    subset,
    shape=None,
    color=f"{mapping_gene}_expression",
    size=10,
    cmap="Reds",  # optional: you can change color map
    title=f"Patient ID: {patient_id}, Run Id: {cell_ID_like} PMN Count: {pmn_count} Gene: {mapping_gene}"
)
```
